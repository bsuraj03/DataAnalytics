```python
data = {
    "Name": ["Barney", "Lily", "Marshall", "Robin", "Ted"],
    "Dept.": ["Sales", "Education", "Law", "Media & Entertainment", "Architecture"],
    "Salary": [110000, 36000, 700000, 55000, 42000],
    "Experience": [7,5,3,9,6]
}
```

```python
df = pd.DataFrame(data)
```
